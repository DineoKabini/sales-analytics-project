```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np

df = pd.read_csv("../data/cleaned/clean_train.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum().reset_index()
monthly["month_num"] = range(len(monthly))

X = monthly[["month_num"]]
y = monthly["Sales"]

model = LinearRegression()
model.fit(X, y)

future = np.array([[len(monthly)], [len(monthly)+1], [len(monthly)+2]])
prediction = model.predict(future)

print("Next 3 months forecast:", prediction)
```
